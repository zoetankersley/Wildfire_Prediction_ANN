```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
import pandas as pd
import glob

data = sorted(glob.glob("../data/*.csv"))
print(data)
df = pd.concat((pd.read_csv(f, low_memory=False) for f in data), ignore_index=True)
df.shape
```

```python
df.head()
```

```python
pd.set_option("display.max_rows", 400)
print(df["FIRE_SIZE_CLASS"].value_counts().sort_index())   # target balance
print(df.dtypes.value_counts())                            # how many numeric vs text columns
df.isna().mean().sort_values(ascending=False).head(40)     # most-missing columns
```

```python
drop = [
    # Known only after the fire
    "CONT_DATE", "CONT_DOY", "CONT_TIME",
    "NWCG_CAUSE_CLASSIFICATION", "NWCG_GENERAL_CAUSE", "NWCG_CAUSE_AGE_CATEGORY",
    "MTBS_ID", "MTBS_FIRE_NAME", "ICS_209_PLUS_INCIDENT_JOIN_ID", "ICS_209_PLUS_COMPLEX_JOIN_ID",
    "COMPLEX_NAME",
    # IDs and names (no predictive meaning)
    "FOD_ID", "FPA_ID", "SOURCE_SYSTEM_TYPE", "SOURCE_SYSTEM", "NWCG_REPORTING_AGENCY",
    "NWCG_REPORTING_UNIT_ID", "NWCG_REPORTING_UNIT_NAME", "SOURCE_REPORTING_UNIT",
    "SOURCE_REPORTING_UNIT_NAME", "LOCAL_FIRE_REPORT_ID", "LOCAL_INCIDENT_ID",
    "FIRE_CODE", "FIRE_NAME",
]
df = df.drop(columns=drop, errors="ignore")
df.shape
```

```python
df["DISCOVERY_DATE"] = pd.to_datetime(df["DISCOVERY_DATE"], errors="coerce")
df["DISCOVERY_MONTH"] = df["DISCOVERY_DATE"].dt.month
```

```python
missing = df.isna().mean()
print(missing[missing > 0.5].sort_values(ascending=False))   # look at this list first

df = df.loc[:, missing <= 0.5]                  # drop columns that are more than half empty
df = df.dropna(subset=["FIRE_SIZE_CLASS", "LATITUDE", "LONGITUDE"])   # rows must have a target and location
df.shape
```

```python
print(df["FIRE_SIZE_CLASS"].value_counts(normalize=True).sort_index().round(4))
```

```python
merge = {"A": "A", "B": "B", "C": "C", "D": "D-E", "E": "D-E", "F": "F-G", "G": "F-G"}
df["TARGET"] = df["FIRE_SIZE_CLASS"].map(merge)
df["TARGET"].value_counts()
```

```python
def find(*words):
    return [c for c in df.columns if any(w.lower() in c.lower() for w in words)]

print("WEATHER:", find("tmmx", "tmmn", "temp", "pr", "rmin", "rmax", "humid", "vs", "wind", "vpd", "erc", "bi", "fm", "pdsi", "drought"))
print("VEGETATION:", find("ndvi", "evt", "land", "cover", "fuel", "veg", "forest", "grass", "shrub"))
print("GEOGRAPHY:", find("elev", "slope", "aspect", "tpi", "lat", "lon"))
```

```python
WEATHER = [
    # Conditions on the day of ignition
    "pr", "tmmn", "tmmx", "rmin", "rmax", "vs", "vpd", "fm100", "fm1000", "bi", "erc",
    # Long-term climate at that place and time of year
    "pr_Normal", "tmmn_Normal", "tmmx_Normal", "rmin_Normal", "rmax_Normal",
    "fm100_Normal", "fm1000_Normal", "bi_Normal", "vpd_Normal", "erc_Normal",
    "Annual_precipitation", "Annual_tempreture",
    # How unusual today is compared to history
    "tmmn_Percentile", "tmmx_Percentile", "sph_Percentile", "vs_Percentile",
    "fm100_Percentile", "bi_Percentile", "vpd_Percentile", "erc_Percentile",
    # Season
    "DISCOVERY_DOY", "DISCOVERY_MONTH",
]

VEGETATION = [
    "NDVI-1day", "NDVI_min", "NDVI_max", "NDVI_mean", "MOD_NDVI_12m",
    "CheatGrass", "ExoticAnnualGrass",
    "Land_Cover", "EVT",          # category codes; see below
]

GEOGRAPHY = [
    "LATITUDE", "LONGITUDE",
    "Elevation", "Slope", "TPI", "Aspect",
    "Elevation_1km", "Slope_1km", "TPI_1km", "Aspect_1km",
]
```

```python
from sklearn.model_selection import GroupShuffleSplit

df["REGION"] = (np.floor(df["LATITUDE"]).astype(int).astype(str) + "_" +
                np.floor(df["LONGITUDE"]).astype(int).astype(str))

gss = GroupShuffleSplit(n_splits=1, test_size=0.15, random_state=42)
trval_idx, test_idx = next(gss.split(df, groups=df["REGION"]))
trval, test = df.iloc[trval_idx], df.iloc[test_idx]

gss2 = GroupShuffleSplit(n_splits=1, test_size=0.15 / 0.85, random_state=42)
tr_idx, val_idx = next(gss2.split(trval, groups=trval["REGION"]))
train, val = trval.iloc[tr_idx], trval.iloc[val_idx]

# Sanity checks
print(len(train), len(val), len(test))
print("Region overlap:", set(train.REGION) & set(test.REGION))   # should be empty
pd.DataFrame({s: d["TARGET"].value_counts(normalize=True)
              for s, d in [("train", train), ("val", val), ("test", test)]}).round(3)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

FEATURES = WEATHER + VEGETATION + GEOGRAPHY
num = [c for c in FEATURES if pd.api.types.is_numeric_dtype(df[c])]

imputer = SimpleImputer(strategy="median").fit(train[num])
scaler = StandardScaler().fit(imputer.transform(train[num]))

def prep(d):
    return scaler.transform(imputer.transform(d[num]))

X_train, X_val, X_test = prep(train), prep(val), prep(test)
y_train, y_val, y_test = train["TARGET"], val["TARGET"], test["TARGET"]
```
